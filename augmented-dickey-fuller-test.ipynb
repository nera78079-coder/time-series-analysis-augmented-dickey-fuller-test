```python
import matplotlib.pyplot as plt
import numpy as np
from statsmodels.tsa.stattools import adfuller
```

```python
def generate_ar_process (lags,coefs,length):

    coefs=np.array(coefs)

    series=[np.random.normal() for _ in range(lags)]

    for _ in range(length):

        prev_vals=series[-lags:][::-1]

        new_val=np.sum(np.array(prev_vals)* coefs)+ np.random.normal()

        series.append(new_val)

    return np.array(series)

```

```python
def perform_adf__test (series):
    reseult=adfuller(series)
    print("ADF  Statistic: %f" % reseult[0])
    print("p-value: %f " % reseult[1])
```

<h3>Ar (1) Process </h3>

<h5> 1.Stationary</h5>

```python
ar_1_process=generate_ar_process(1,[.5],100)
ar_1_process
```

```python
plt.figure(figsize=(10,4))
plt.plot(ar_1_process)
plt.title("Stationary AR(1) Process",fontsize=18)
plt.show()
```

```python
perform_adf__test(ar_1_process)
```

<h5> 2.Non-Stationary</h5>

```python
ar_1_process_unit_root=generate_ar_process(1,[1],100)
plt.Figure(figsize=(10,4))
plt.plot(ar_1_process_unit_root)
plt.title("Non-Stationary AR(1) Process")
plt.show()
```

```python
perform_adf__test(ar_1_process_unit_root)
```

<h3> AR(2) Process</h3>

<h4> 1.Stationary </h4>

```python
ar_2_process=generate_ar_process(2,[.5,.3],100)
plt.figure(figsize=(10,4))
plt.plot(ar_2_process)
plt.title("Stationary AR(2) Process")
plt.show()
```

```python
perform_adf__test(ar_2_process)
```

```python
perform_adf__test(ar_2_process_unit_root)
```
